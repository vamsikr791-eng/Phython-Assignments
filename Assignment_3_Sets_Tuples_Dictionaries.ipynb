```python
# Q1 Phone Contact Book

contacts = {}

while True:
    print("\nPhone Contact Book")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Show All Contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter contact name: ")

        if name in contacts:
            print("Contact already exists")
        else:
            phone = input("Enter phone number: ")
            contacts[name] = phone
            print("Contact added successfully")

    elif choice == "2":
        name = input("Enter contact name to search: ")

        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts available")
        else:
            print("All Contacts:")
            for name, phone in contacts.items():
                print(name, ":", phone)

    elif choice == "4":
        print("Exiting Contact Book")
        break

    else:
        print("Invalid choice")
```

```python
# Q3 Student Marks Dictionary

marks = {
    "Aman": 82,
    "Priya": 91,
    "Rahul": 45,
    "Sneha": 67,
    "Kiran": 30
}

passed_students = {}

topper = ""
highest_marks = 0

for name, mark in marks.items():
    if mark >= 40:
        print(name, "- Pass")
        passed_students[name] = mark
    else:
        print(name, "- Fail")

    if mark > highest_marks:
        highest_marks = mark
        topper = name

print("Topper:", topper)
print("Topper Marks:", highest_marks)
print("Passed Students:", passed_students)
```

```python
# Q4 Common Friends

riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

mutual_friends = riya_friends.intersection(arjun_friends)

only_riya = riya_friends.difference(arjun_friends)

all_friends = riya_friends.union(arjun_friends)

suggestions = arjun_friends.difference(riya_friends)

print("Mutual Friends:", sorted(mutual_friends))
print("Only Riya's Friends:", sorted(only_riya))
print("All Friends:", sorted(all_friends))
print("Friend Suggestions for Riya:", sorted(suggestions))
```

```python
# Q5 Character Counter

text = input("Enter text: ")

char_count = {}

for char in text:
    if char != " ":
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1

print("Character count:", char_count)

most_frequent = ""
highest_count = 0

for char, count in char_count.items():
    if count > highest_count:
        highest_count = count
        most_frequent = char

if highest_count > 0:
    print("Most frequent character:", most_frequent,
          "(", highest_count, "times)")
else:
    print("No characters entered")
```

```python
# Q6 Restaurant Menu Billing

menu = {
    "dosa": 60,
    "idli": 40,
    "vada": 30,
    "coffee": 25,
    "tea": 15
}

total = 0

print("Restaurant Menu:")
for item, price in menu.items():
    print(item, "- Rs.", price)

while True:
    order = input("Enter item (or 'bill' to finish): ").lower().strip()

    if order == "bill":
        break
    elif order in menu:
        total += menu[order]
        print(order, "added. Price: Rs.", menu[order])
    else:
        print("Item not available")

print("Total before discount: Rs.", total)

if total > 200:
    discount = total * 0.10
    total = total - discount
    print("Discount applied: Rs.", discount)

print("Final Bill: Rs.", total)
```

```python
# Q7 City Coordinates (Tuples)

cities = [
    ("Hyderabad", 17.38, 78.48),
    ("Mumbai", 19.07, 72.87),
    ("Chennai", 13.08, 80.27),
    ("Delhi", 28.61, 77.20)
]

print("City Coordinates:")

for city, latitude, longitude in cities:
    print(city, "- Latitude:", latitude, "Longitude:", longitude)

north_city = ""
highest_latitude = -90

for city, latitude, longitude in cities:
    if latitude > highest_latitude:
        highest_latitude = latitude
        north_city = city

print("Farthest North City:", north_city)
print("Highest Latitude:", highest_latitude)

# Trying to modify Mumbai's latitude:
# cities[1][1] = 20.00
# Error: TypeError: 'tuple' object does not support item assignment
# Reason: Tuples are immutable, so their values cannot be changed.
```

```python
# Q8 Library Book Tracker

books = {
    "Python Basics": 3,
    "Data Science 101": 0,
    "Machine Learning": 2,
    "SQL Guide": 1
}

book_name = input("Enter book name to borrow: ")

if book_name not in books:
    print("Book not found")

elif books[book_name] == 0:
    print("Currently unavailable")

else:
    books[book_name] -= 1
    print("Book issued successfully")

print("Updated books:", books)
```

```python
# Q9 Group Employees by Department

employees = [
    ("Ravi", "IT"),
    ("Anjali", "HR"),
    ("Suresh", "IT"),
    ("Meena", "Finance"),
    ("Arun", "HR"),
    ("Divya", "IT")
]

departments = {}

for name, dept in employees:
    if dept not in departments:
        departments[dept] = []

    departments[dept].append(name)

print("Employees grouped by department:")
print(departments)

print("\nEmployee count by department:")

for dept, names in departments.items():
    print(dept, ":", len(names))
```

```python
# Q10 Quiz Answer Checker

answer_key = ("B", "C", "A", "D", "B")

wrong_questions = set()
score = 0

for i in range(5):
    answer = input(f"Enter answer for Question {i + 1}: ").upper().strip()

    if answer == answer_key[i]:
        print("Question", i + 1, "- Correct")
        score += 1
    else:
        print("Question", i + 1, "- Wrong")
        wrong_questions.add(i + 1)

print("\nWrong Question Numbers:", sorted(wrong_questions))
print("Final Score:", score, "/ 5")

if score == 5:
    print("Remark: Excellent")
elif score >= 3:
    print("Remark: Good")
else:
    print("Remark: Needs improvement")
```
